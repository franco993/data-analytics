```python
avg_Salary_by_EX_lvl=df2.groupby('experience_level')['salary_usd'].mean().round(2)
print(avg_Salary_by_EX_lvl)
y=avg_Salary_by_EX_lvl.values
x=avg_Salary_by_EX_lvl.index
```

```python
plt.figure(figsize=(10,6))
plt.title("Average salary by experience level")
plt.bar(x,y,color="#abc4ff",edgecolor='black', width=0.6)
plt.xlabel("experience level")
plt.ylabel("Salary (USD)")
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
remote_by_employment = df2.groupby('employment_type')['remote_ratio'].mean().round(2)
print(remote_by_employment)
x=remote_by_employment.index
y=remote_by_employment.values
```

```python
plt.figure(figsize=(10,6))
plt.title("Average remote ratio by employment type")
plt.bar(x,y,color='#c1d3fe',edgecolor="black")
plt.xlabel("Employment type",fontsize=10)
plt.ylabel("Avg Remote ration",fontsize=10)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.grid(axis='y',linestyle='--',alpha=0.5)
plt.show()
```

```python
avg_salary_by_experience_years=df2.groupby('years_experience')['salary_usd'].mean()
x=avg_salary_by_experience_years.index
y=avg_salary_by_experience_years.values
```

```python
plt.figure(figsize=(10,6))
plt.title("Relation between year experience and salary (USD)")
plt.scatter(x,y,color='#194a7a')
plt.xlabel("year of experience")
plt.ylabel("Avg salary (USD)")
plt.grid()
plt.show()
```

```python
avg_salary_by_education=df2.groupby('education_required')['salary_usd'].mean()
x=avg_salary_by_education.index
y=avg_salary_by_education.values
```

```python
plt.figure(figsize=(10,6))
df2.boxplot(column="salary_usd", by="education_required")
# plt.title("Salary distribution by education level")
plt.xlabel("Education Level")
plt.ylabel("Salary (USD)")
plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
Avg_salary_by_country=df2.groupby('company_location')['salary_usd'].mean().round(2)
# print(Avg_salary_by_country)
x=Avg_salary_by_country.index
y=Avg_salary_by_country.values
```

```python
plt.figure(figsize=(10,6))
plt.title('Avg Salary by Company location')
plt.bar(x,y,color='#7593af')
plt.xticks(rotation=60,fontsize=10)
plt.yticks(fontsize=10)
plt.xlabel("company locations")
plt.ylabel("Salary (USD)")
ax=plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.grid(True,axis='y',linestyle='--',alpha=0.5)
plt.show()
```

```python
df2['posting_date'] = pd.to_datetime(df2['posting_date'], errors='coerce')
df2['year_month'] = df2['posting_date'].dt.to_period('M')
avgJopLenthByDate=df2.groupby('year_month')['job_description_length'].mean().round(2)
print(avgJopLenthByDate)
x=avgJopLenthByDate.index.astype(str)
y=avgJopLenthByDate.values
x.unique
```

```python
plt.figure(figsize=(10,6))
plt.title('Average jop describtion length by posting date')
plt.plot(x,y,color='#00202e',marker='.')
plt.xlabel('Posting Date')
plt.ylabel('Jop Description Length')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
industry_des=df2['industry'].value_counts()
x=industry_des.index
y=industry_des.values
explode=[0.6,0.5,0.4,0.4,0.3,0.3,0.3,0.2,0.2,0.2,0.2,0.1,0,0,0]
colors = sns.color_palette("pastel", len(x))
plt.pie(y,labels=x,autopct='%1.1f%%',explode=explode,colors=colors)
plt.show()
```
